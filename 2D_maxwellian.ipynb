```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = 1.0
T = 1.0
kB = 1.0
```

```python
# r = np.linspace(0, 5, 30)
# theta = np.linspace(0, 360, 30)

N = 1000

vx = np.linspace(-10, 10, N)
vy = np.linspace(-10, 10, N)


VX, VY = np.meshgrid(vx, vy)
```

```python
V = np.sqrt(VX**2 + VY**2)
```

```python
f = m*V/(kB*T) * np.exp(-m*V**2/(2*kB*T))
```

```python
f.shape
```

```python
plt.imshow(f, cmap='plasma')
plt.colorbar()
```

```python
v = np.linspace(0, 10, 1000)
theta = np.linspace(0, 360.0, 1000)

vx = v[:, None] * np.cos(theta[None, :])
vy = v[:, None] * np.sin(theta[None, :])

V = np.sqrt(vx**2 + vy**2)
```

```python
f = m*V/(kB*T) * np.exp(-m*V**2/(2*kB*T))
```

```python
plt.imshow(f)
```

```python
v     = np.linspace(0, 10, 300)
theta = np.linspace(0, 2*np.pi, 300)   # radians, endpoint included so the disc closes

VX = v[:, None] * np.cos(theta[None, :])
VY = v[:, None] * np.sin(theta[None, :])

# evaluate on the radius, then broadcast across theta
f  = m/(2*np.pi*kB*T) * np.exp(-m*(v-3)**2/(2*kB*T))
F  = np.broadcast_to(f[:, None], VX.shape)

plt.pcolormesh(VX, VY, F, shading='gouraud')
plt.gca().set_aspect('equal')
plt.colorbar(label='f(vx, vy)')
```
